# **Tokenization :**

Tokenization is the process of breaking text into smaller units called tokens. A token can be a word, punctuation mark, number, or special symbol.

## **Tokenization using SpaCy**

```python
# Then impor spaCy:
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
# 2. Basic Tokenization
text = "Programming language is eassy to learn."
doc = nlp(text)
for token in doc:
  print(token.text)
```

```python
# 3. Tokenization with Token Index
# Each token has an indez indicating its position in the document
# token.i gives the position/index of the token
text = "Programming language is eassy to learn."
doc = nlp(text)
for token in doc:
  print(token.i, token.text)
```

```python
# Tokenization with Token Attributes
# SpaCy provides several userful attributes for each token
text = "Machile Learning is eassy to learn."
doc = nlp(text)
for token in doc:
  print(
      token.text,
      token.is_alpha,
      token.is_punct,
      token.is_space
  )
```

```python
# 5. Tokenization of a sentence with Punctuation
text = "Hii, What is Python? you know."
doc = nlp(text)
for token in doc:
  print(token.text)
```

```python
# 6. Tokenization of Numbers
text = "My company's total expense in $50000."
doc = nlp(text)
for token in doc:
  print(token.text)
```

SpaCy reconizes numbers and punctuation separately.

```python
# 7. Tokenization of Email Addresses.
text = "My email is adeshksingh6393@gmail.com"
doc = nlp(text)
for token in doc:
  print(token.text)
```

```python
# 8. Tokenization of URLs
text = "Visit my github https://github.com/Adeshksingh01 for more information."
doc = nlp(text)
for token in doc:
  print(token.text)
```

```python
# 9. Tokenization of Contractions
text = "I cann't do this work because i don't like."
doc = nlp(text)
for token in doc:
  print(token.text)
```

```python
# 10. Tokenization of Hyphenated words
text = "My favorite subject is ML-Pipeline."
doc = nlp(text)
for token in doc:
  print(token.text)
```

```python
# 11. Removing Punctuation After Tokenization
text = "Rahul is smart, good, and intelligent!"
doc = nlp(text)
for token in doc:
  if not token.is_punct:
    print(token.text)
```

```python
# 12. Getting Only Alphabetic Token
# We use is_alpha to select only alphabetic tokens
text = "Alto 800 is easy to drive!"
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text)
```

```python
# 13. Converting Tokens into a List
text = "Alto is cheapest car."
doc = nlp(text)
tokens = [token.text for token in doc]
print(tokens)
```

```python
# Optional way to convert token into a list
text = "Alto is cheapest car."
doc = nlp(text)
tokens = []
for token in doc:
  tokens.append(token.text)
print(tokens)
```

```python
# 14. Tokenization of Multiple Sentences
text = "My sefl AKS. I am from UP."
doc = nlp(text)
for token in doc:
  print(token.text)
```

SpaCy processes the entire text as a Doc object.

```python
# 15. Tokenization and Sentence Segmentation
# SpaCy can also identify individual sentences
text = "My self AKS. I am from UP."
doc = nlp(text)
for sentence in doc.sents:
  print("Sentence:", sentence.text)
  for token in sentence:
    print(token.text)

print()
```

## **Tokenization using NLTK**

```python
# Install NLTK
!pip install nltk
```

```python
# Import nltk
import nltk
```

```python
# Depending on your NLTK version, download the required tokenizer resources:
nltk.download('punkt_tab')
```

```python
# 1. Word Tokenization
from nltk.tokenize import word_tokenize
text = "Python is easy comparision of C, C++ and Java."
tokens = word_tokenize(text)
print(tokens)
```

- Word tokenize(text) breaks the sentence into individual tokens.

```python
# 2. Sentence Tokenization
from nltk.tokenize import sent_tokenize
text = "NLTK is useful for NLP. NLP is Stand for Natural Process Language. NLTK is the part of Python."
sentence = sent_tokenize(text)
print(sentence)
```

- sent_tokenize(text) is useful when we want to process a document sentence by sentence.

```python
# 3. Word Tokenization of Multiple Sentences
from nltk.tokenize import word_tokenize
text = """She always runs faster than anyone else in her class.
Several wolves howled at the moon last night.
Are you feeling happier now?"""

tokens = word_tokenize(text)
print(tokens)
```

```python
# 4. Tokenizing User Input
from nltk.tokenize import word_tokenize
text = input("Enter a sentence:")
tokens = word_tokenize(text)
print("Tokens:",tokens)
```

```python
# 5. Tokenization with Punctuation
text = "Hii, who are you? what is your name!"
tokens = word_tokenize(text)
print(tokens)
```

```python
# 6. Tokenization of Numbers
text = "Alto 800 is 500000 rupees."
tokens = word_tokenize(text)
print(tokens)
```

```python
# 7. Tokenization with Contractions
text = "I cann't purchase Alto, I don't like it."
tokens = word_tokenize(text)
print(tokens)
```

```python
# 8. Sentence Tokenization + Word Tokenization
from nltk.tokenize import word_tokenize, sent_tokenize
text = """She always runs faster than anyone else in her class.
Several wolves howled at the moon last night.
Are you feeling happier now?"""

sentences = sent_tokenize(text)
for sent in sentences:
  words = word_tokenize(sent)
  print(words)
```

```python
# 9. Count Number of Tokens
from nltk.tokenize import word_tokenize
text = "She always runs faster than anyone else in her class."
tokens = word_tokenize(text)
print("Tokens:", tokens)
print("Number of Tokens:", len(tokens))
```

```python
# 10. Count only words
import string
text = "She always, runs faster than anyone else in her class!"
tokens = word_tokenize(text)
words = [token for token in tokens if token not in string.punctuation]

print("All Tokens:", tokens)
print("Words:", words)
print("Number of words:", len(words))
```

```python
# 11. Convert Tokens to Lowercase
text = "My name is ADESH and my full name is Adesh Kumar Singh."
tokens = word_tokenize(text)
lower_tokens = [word.lower() for word in tokens]

print("Original Tokens:")
print(tokens)
print("\nLowercase Tokens:")
print(lower_tokens)
```

```python
# 12. Tokenization using wordpunct_tokinize()
from nltk.tokenize import wordpunct_tokenize
text = "Hey, are you feeling happier now?"
tokens = wordpunct_tokenize(text)
print(tokens)
```

```python
# 13. Tokenizing a Paragraph
from nltk.tokenize import word_tokenize, sent_tokenize
text = """
The children were running through the fallen leaves in the forest.
Earlier, they had seen three mice hiding behind the bushes.
One child swam across the freezing river, while the others threw stones.
"""

sentences = sent_tokenize(text)
print("Number of sentences:", len(sentences))

for i, sentence in enumerate(sentences, start = 1):
  print(f"\nsentence {i}:")
  print(sentence)

  tokens = word_tokenize(sentence)
  print("Tokens:", tokens)
```

```python
# 15. Complete Tokenization Program
import nltk
import string
from nltk.tokenize import sent_tokenize, word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')

text = """
The children were running through the fallen leaves in the forest.
Earlier, they had seen three mice hiding behind the bushes.
One child swam across the freezing river, while the others threw stones.
"""

# Step 1: Sentence Tokenization
sentences = sent_tokenize(text)

print("Sentences:")
for sentence in sentences:
  print(sentence)

# Step 2: Word Tokenization
tokens = word_tokenize(text)
print("\nAll Tokens:")
print(tokens)

# Step 3: Remove punctuation
words = [word for word in tokens if word not in string.punctuation]
print("\nTokens without punctuation:")
print(words)

# Step 4: Convert to lowercase
lower_words = [word.lower() for word in words]
print("\nLowercase Tokens:")
print(lower_words)

# Step 5: Count tokens
print("\nTotal tokens:", len(tokens))
print("\nTotal words:", len(words))
```
