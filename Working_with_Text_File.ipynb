```python
# Opens 'practice.txt' in default read mode and stores the variable named my_file.
my_file = open('practice.txt')
```

```python
# Reads the entire content of the file and returns it as a single string
my_file.read()
```

```python
# Resets the file cursor (pointer) back to the beginning of the file
my_file.seek(0)
```

```python
# Safely opens the file, extracts the first line from the list of all lines, and prints it
with open('practice.txt', 'r') as txt:
  first_line = txt.readlines()[0]
print(first_line)
```

```python
# Safely opens the file and efficiently iterates through it line by line, printing each without adding extra newlines
with open('practice.txt', 'r') as txt:
  for line in txt:
    print(line, end='')
```

```python
# Closes the file safely and frees up system resources
my_file.close()
```
