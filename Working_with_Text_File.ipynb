## Advanced Operations

```python
# 1. Creating and Writing to a Text File

# Open file in write mode
file = open('employee.txt', 'w')

file.write('Riya\n')
file.write('Salmaan\n')
file.write('Amitabh\n')
file.write('Sunil\n')

file.close()
print('File created successfully.')
```

```python
# 2. Writing Multiple Lines Using Writelines()
employees = [
    "Keshav\n",
    "Ayushi\n",
    "Rahul\n",
    "Abhishek\n"
]

with open('employee.txt', 'w') as file:
  file.writelines(employees)

print("Data written successfully.")
```

```python
# 3. Reading the Entire File
with open('employee.txt', 'r') as file:
  data = file.read()

print(data)
```

```python
# 4. Reading File Line by Line
with open('employee.txt', 'r') as file:
  for line in file:
    print(line.strip()) # Print(line)
```

- .strip() remove the newline character (\n) from each line.

```python
# 5. Using readline
with open('employee.txt', 'r') as file:
  first_line = file.readline()
  second_line = file.readline()
  third_line = file.readline()

print("First Line:", first_line.strip())
print("Second Line:", second_line.strip())
print("Third Line:", third_line.strip())
```

```python
# 6. Using readlines()
with open('employee.txt', 'r') as file:
  all_lines = file.readlines()

print(all_lines)
```

```python
# We can also process the list
# Using readlines() for print context into line by line
with open('employee.txt', 'r') as file:
  lines = file.readlines()

for line in lines:
  print(line.strip())
```

```python
# 7. Appending Data to a File.
with open('employee.txt', 'a') as file:
  file.write('Shivam\n')
  file.write('Sakshi\n')
  file.write('Adesh\n')

print("New students added.")
```

```python
# 8. Writing User Input to File.
name = input("Enter your name:")
age = input("Enter your age:")

with open('clients.txt', 'a') as file:
  file.write("Name: " + name + "\n")
  file.write("Age: " + age + "\n")

print("Client information saved.")
```

```python
# 9. Counting Number of Lines.
with open('employee.txt', 'r') as file:
  all_line = file.readlines()
print("Number of Lines:", len(all_line))
```

```python
# 10. Counting Words in a Text File.
# .split() is split character after space
with open("employee.txt", 'r') as file:
  data = file.read()
words = data.split()

print("Number of words:", len(words))
```

```python
# 11. Counting Characters
# To include spaces
with open('employee.txt', 'r') as file:
  data = file.read()

print("Number of characters:", len(data))
```

```python
# To exclude spaces:
characters = data.replace(' ', '').replace('\n', '')
print('Characters excluding spaces:', len(characters))
```

```python
# 12. Searching for word
search_word = input("Enter word to search:")

with open('employee.txt', 'r') as file:
  data = file.read()

if search_word.lower() in data.lower():
  print('Word found.')
else:
  print('Word not found.')
```

```python
# 13. Counting Occurrences of a word
search_word = input("Enter word:")

with open('employee.txt', 'r') as file:
  data = file.read()

count = data.lower().count(search_word.lower())
print('Occurrences:', count)
```

```python
# 14. Copying contents from one file to another.
with open('employee.txt', 'r') as file:
  data = file.read()

with open('coying.txt', 'w') as copy:
  copy.write(data)

print('File copied successfully.')
```

```python
# 15. Reading only the first 10 characters
with open('employee.txt', 'r') as file:
  data = file.read(10)

print(data)
```
